```python
# Saving yield for Combine
mxx_map = {"200GeV": "1", "500GeV": "2",  "800GeV": "3", "1000GeV": "4"}
mzd_map = {"0p25GeV": "1", "1p2GeV": "2", "5p0GeV": "3"}
# var = "x" # 

vs = "38" # for saving the filename and signal point

groups = [
    fmulxy1,
    fmulxy2,
    fmulxy3,
    fmulxy4,
    fmulxy5,
    fmulxy6,
    fmulxy7,
    fmulxy8,
    fmulxy9,
    fmulxy0,
    fmulxyi,
    fmulxyx,
]

with open(f"combine_inputs_{vs}.txt", "w") as fout:

    for group in groups:

        for ctau_idx, sample in enumerate(group, start=1):

            yld = output[sample]['cutflow'][ch].rows[cut]['weighted']

            _, mxx, mzd, _ = sample.split("_")

            identifier = (
                mxx_map[mxx]
                + mzd_map[mzd]
                + str(ctau_idx)
            )

            fout.write(f"_fmu_{vs}_{identifier} {yld:.2f}\n")
```
